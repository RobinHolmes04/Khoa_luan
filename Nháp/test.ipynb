```python
import pandas as pd
from scipy import stats

df = pd.read_csv("Dữ liệu ví dụ & thực hành t-test bắt cặp, độc lập và ANOVA - Dữ liệu DLS thực hành.csv")
df
```

```python
df_cleaned = df.dropna()
df_cleaned["INR sau"]
```

```python
def kiem_tra_chuan_shapiro(data_series):
    #chạy kiểm định
    stat, p_value = stats.shapiro(data_series)
    
    # Đánh giá kết quả dựa trên p-value
    is_normal = p_value > 0.05
    
    if is_normal:
        print(f"Dữ liệu có phân phối chuẩn (p = {p_value:.4f})")
    else:
        print(f"Dữ liệu KHÔNG có phân phối chuẩn (p = {p_value:.4f})")
        
    return is_normal, p_value
```

```python
def kiem_dinh_ttest_bat_cap(truoc,sau):
    stat,p_value = stats.ttest_rel(truoc,sau)
    if p_value < 0.05:
        print("Có liên quan")
    else:
        print("Không liên quan")
    return stat,p_value
```
