```python
def independent_variable_d_or_c(df, ten_cot, nguong_discrete=10):
    # 1. Trích xuất cột và loại bỏ dữ liệu trống để không bị lỗi
    cot_du_lieu = df[ten_cot].dropna()
    
    # Nếu toàn bộ cột bị trống
    if cot_du_lieu.empty:
        return "Không xác định (Cột rỗng)"
        
    # 2. Kiểm tra bản chất kiểu dữ liệu của Pandas
    if not pd.api.types.is_numeric_dtype(cot_du_lieu):
        # Không phải là số (Ví dụ: "Nam"/"Nữ", "Nhẹ"/"Nặng") -> Chắc chắn là Discrete
        return "discrete"
        
    # 3. Nếu là số, kiểm tra số lượng giá trị duy nhất (Unique values)
    so_gia_tri_khac_biet = cot_du_lieu.nunique()
    
    if so_gia_tri_khac_biet <= nguong_discrete:
        # Số lượng giá trị quá ít (ví dụ 1, 2, 3, 4) -> Discrete (Biến phân loại)
        return "discrete"
    else:
        # Số lượng giá trị phong phú -> Continuous (Biến định lượng)
        return "continuous"
```
